```python
train = train.drop(['PassengerId', 'Name', 'Cabin', 'Ticket'], axis=1)
```

```python
# train 데이터의 null을 찾아서 평균으로 바꾸기
mean1 = train['Age'].mean()
train['Age'] = train['Age'].fillna(mean1)
```

```python
train.isna().any()
```

```python
# train 데이터의 null을 찾아서 평균으로 바꾸기
mean2 = test['Age'].mean()
test['Age'] = test['Age'].fillna(mean2)
```

```python
test.isna().any()
```

```python
mean3 = train['Fare'].mean()
train['Fare'] = train['Fare'].fillna(mean3)
```

```python
mean4 = test['Fare'].mean()
test['Fare'] = test['Fare'].fillna(mean4)
```

```python
train.isna().any()
```

```python
test.isna().any()
```

```python
mode1 = train['Embarked'].mode()[0]
train['Embarked'] = train['Embarked'].fillna(mode1)
```

```python
train.isna().any()
```

```python
# 더미변수처리
train = pd.get_dummies(train, columns=['Sex', 'Embarked'],
                      drop_first = True)
test = pd.get_dummies(test, columns=['Sex', 'Embarked'],
                      drop_first = True)
```

```python
train[0:3]
```

```python
train = train.drop(['Sex_male', 'Embarked_Q', 'Embarked_S'], axis=1)
test = test.drop(['Sex_male', 'Embarked_Q', 'Embarked_S'], axis=1)
```

```python
train[0:3]
```

```python
train.to_csv('preprocessed_train.csv', index=False)
```

```python
x_train, y_train = train.drop(['Survived'], axis=1), train.Survived.values
```

```python
from scipy.stats import pearsonr
```

```python
corrs = []
for col in x_train.columns:
    corr = pearsonr(x_train[col], y_train)[0]
    print('{} : {}'.format(col, corr))
    corrs.append([col, corr])
```

```python
corrs = pd.DataFrame(corrs, columns = ['columns', 'correlation'])
```

```python
corrs.sort_values('correlation')
```

```python
corrs['correlation'] = corrs['correlation'].apply(np.abs)
```

```python
corrs.sort_values('correlation')
```

```python
corrs.loc[corrs['correlation'].nlargest(3).index]
```

```python
corrs.loc[corrs['correlation'].nlargest(3).index]['columns'].values
```

```python
selected = corrs.loc[corrs['correlation'].nlargest(3).index]['columns'].values
```

```python
selected
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.metrics import accuracy_score
```

```python
rfc = RandomForestClassifier(random_state=42)
scores = cross_val_score(rfc, x_train, y_train, scoring = 'accuracy',
                        cv = 10)
```

```python
print(scores)
print('='*20)
print(np.mean(scores))
```

```python
rfc = RandomForestClassifier(random_state=42)
scores = cross_val_score(rfc, x_train[selected], y_train, scoring = 'accuracy',
                        cv = 10)
```

```python
print(scores)
print('='*20)
print(np.mean(scores))
```

```python
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import chi2
```

```python
bestfeatures = SelectKBest(score_func = chi2, k = 3)
```

```python
fit = bestfeatures.fit(x_train, y_train)
```

```python
dfscores = pd.DataFrame(fit.scores_)
dfcolumns = pd.DataFrame(x_train.columns)
```

```python
dfscores
```

```python
featureScores = pd.concat([dfcolumns, dfscores], axis=1)
featureScores.columns = ['name', 'score']
```

```python
featureScores
```

```python
featureScores.nlargest(4, 'score')
```

```python
x_train_chi2 = bestfeatures.fit_transform(x_train, y_train)
```

```python
rfc1 = RandomForestClassifier(random_state=42)
scores = cross_val_score(rfc1, x_train, y_train, scoring = 'accuracy',
                        cv = 10)
print(scores)
print('='*20)
print(np.mean(scores))
```

```python
rfc2 = RandomForestClassifier(random_state=42)
scores = cross_val_score(rfc2, x_train[selected], y_train, scoring = 'accuracy',
                        cv = 10)
print(scores)
print('='*20)
print(np.mean(scores))
```

```python
rfc3 = RandomForestClassifier(random_state=42)
scores = cross_val_score(rfc3, x_train_chi2, y_train, scoring = 'accuracy',
                        cv = 10)
print(scores)
print('='*20)
print(np.mean(scores))
```
