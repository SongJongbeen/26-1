```python
from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score
```

```python
dt = pd.read_csv('M5__train_cleansed2.csv')
```

```python
dt.head(3)
```

```python
dt[0:3]
```

```python
x_train, x_test, y_train, y_test = train_test_split(dt.drop('Survived', axis=1),
                                                           dt['Survived'].values,
                                                           test_size=20,
                                                           random_state=3)
```

```python
x_train[0:3]
```

```python
x_test[0:3]
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
pd.DataFrame([model.feature_importances_],
             columns = x_train.columns)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
parameters = {'max_depth': np.arange(1,10)}
```

```python
tree = GridSearchCV(DecisionTreeClassifier(),
                    parameters).fit(x_train, y_train)
```

```python
y_pred = tree.predict(x_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
parameters = {'max_depth': np.arange(1,10),
              'criterion': ['gini', 'entropy']}
```

```python
grid_search = GridSearchCV(DecisionTreeClassifier(random_state=0),
                           parameters).fit(x_train, y_train)
```

```python
y_pred = grid_search.predict(x_test)
```

```python
accuracy_score(y_test, y_pred)
```
