```python
df6
```

```python
df7
```

```python
pd.merge(df6, df7, on = 'name')
```

```python
pd.merge(df6, df7, on = 'name', how = 'inner')
```

```python
pd.merge(df6, df7, on = 'name', how = 'left')
```

```python
pd.merge(df6, df7, on = 'name', how = 'outer')
```
